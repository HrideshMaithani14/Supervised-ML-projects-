```python
import seaborn as sns
import pandas as pd
import numpy as np
import sys
```

```python
df = pd.read_csv('car_price_svr_svc_dataset.csv')
df.head()
```

```python
X=df.drop('expensive',axis=1)
y=df['expensive']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.svm import SVR
svr = SVR()
svr.fit(x_train,y_train)
ypred = svr.predict(x_test)
```

```python
from sklearn.metrics import r2_score ,mean_absolute_error,accuracy_score
print(r2_score(ypred,y_test))
print(mean_absolute_error(ypred,y_test))
```

```python
from sklearn.model_selection import GridSearchCV
parems ={
    'C':[0.1,1.0,0.10,100,1001],
    'gamma':[1,0.1,0.01,0.02],
    'kernel':['rbf']
}

grid = GridSearchCV(svr,param_grid=parems,refit=True,verbose=3)
```

```python
grid.fit(x_train,y_train)
y_pred = grid.predict(x_test)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_auc_score
```

```python
print('accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_auc_score')
print(r2_score(y_pred,y_test))
print(mean_absolute_error(y_pred,y_test))
```

```python
print(grid.best_params_)
print(grid.best_score_)
```
