# Used Car Price Prediction

## 1) Problem statement.

* This dataset comprises used cars sold on cardehko.com in India as well as important features of these cars.
* If user can predict the price of the car based on input features.
* Prediction results can be used to give new seller the price suggestion based on market condition.

## 2) Data Collection.
* The Dataset is collected from scrapping from cardheko webiste
* The data consists of 13 column and 15411 rows.

```python
df.drop(columns=['car_name','brand'],axis=1,inplace=True)
```

```python
X=df.drop('selling_price',axis=1)
y=df['selling_price']
```

```python
X.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
X['model']=le1.fit_transform(X['model'])
```

```python
numerical_feature = X.select_dtypes(exclude='object').columns
car_feature=['seller_type','fuel_type','transmission_type']
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
num_trans=StandardScaler()
car_transfor=OneHotEncoder(drop='first')
preprocess = ColumnTransformer(
    [
        ('onehotencoder',car_transfor,car_feature),
        ('standersclr',num_trans,numerical_feature)
    ]
    
)
```

```python
X=preprocess.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.25,random_state=41)
```

```python
from sklearn.linear_model import LinearRegression ,Ridge,Lasso
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
def Evel_predict(true,predict):
    r2_scor = r2_score(true,predict)
    mae= mean_absolute_error(true,predict)
    mse= mean_squared_error(true,predict)
    rmse = np.sqrt(mse)
    return r2_scor,mae,rmse
```

```python
models={
    'linearRegression':LinearRegression(),
    'SVR':SVR(),
    'KNeighborsRegressor':KNeighborsRegressor(),
    'DecisionTreeRegressor':DecisionTreeRegressor(),
    'RandomForestRegressor':RandomForestRegressor(),
    'AdaBoostRegressor':AdaBoostRegressor()
    }
for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train,y_train)

    #Train/Test prediction
    Y_train_pred =model.predict(X_train)
    Y_test_pred = model.predict(X_test)

    
    r2_scortrain,maetrain,rmsetrain=Evel_predict(y_train,Y_train_pred)
    r2_scortest,maetest,rmsetest=Evel_predict(y_test,Y_test_pred)

    print(list(models.keys())[i])
    print('Train set prediction\n')
    print('r2_score:{:.4f}'.format(r2_scortrain))
    print('mean Absolute Error:{:.4f}'.format(maetrain))
    print('RootMeanSquaredError{:.4f}'.format(rmsetrain))

    print('-'*32)

    print('Test Set Prediction')
    print('r2_score:{:.4f}'.format(r2_scortest))
    print('mean Absolute Error:{:.4f}'.format(maetest))
    print('Root Mean Squared Error:{:.4f}'.format(rmsetest))

    print('='*32)
    print('\n')


    
```

```python
parem_decision_tree={
    'criterion':['squared_error', 'friedman_mse', 'absolute_error'],
    'max_depth': [1,2,3,4,5,6,10],
    'min_samples_split': [0.1,0.2,0.3,1],
    'min_samples_leaf': [0.1,0.2,0.01,1],
    'max_features': [1,2,3,4,5,6]
}
parem_forest={ 
    'criterion': ['squared_error', 'absolute_error', 'friedman_mse', 'poisson'] ,
   'max_depth': [1,2,3,4,5,6],
    'min_samples_split': [0.1,0.2,0.3,1],
    'min_samples_leaf': [0.1,0.2,0.01,1],
    'max_features': [1,2,3,4,5,6]

}
paremadda={
    'n_estimators':[10,20,30,50,60,90,100],
    'learning_rate':[0.1,.2,.3,.5,1],
    'loss':['linear', 'square', 'exponential']
    
}
```

```python
models2=[
    ('DecisionTreeRegressor',DecisionTreeRegressor(),parem_decision_tree),
    ('RandomForestRegressor',RandomForestRegressor(),parem_forest),
    ('AdaBoostRegressor', AdaBoostRegressor(),paremadda)
]
```

```python
from sklearn.model_selection import RandomizedSearchCV
best_parem ={}
for name,model,parem in models2:
    random = RandomizedSearchCV(estimator=model,
                                param_distributions=parem,
                                n_iter=100,
                                cv=3,
                                verbose=2,
                                n_jobs=-1
                                )
    random.fit(X_train,y_train)
    best_parem[name]=random.best_params_
```

```python
for model_name in best_parem:
    print(f'________best parem{model_name}_________')
    print(best_parem[model_name])
```
