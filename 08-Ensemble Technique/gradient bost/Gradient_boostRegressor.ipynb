```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings

warnings.filterwarnings("ignore")

%matplotlib inline
```

```python
df = pd.read_csv('cardekho_imputated.csv',index_col=[0])
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(columns=['car_name','brand'],axis=1,inplace=True)
```

```python
X= df.drop('selling_price',axis=1)
y=df['selling_price']
```

```python
X['model'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
leb = LabelEncoder()
X['model']=leb.fit_transform(X['model'])
```

```python
numerical_feature =X.select_dtypes(exclude='object').columns
categorical_feature = ['seller_type','fuel_type','transmission_type']
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
Numeracl_transformer= StandardScaler()
categorical_transformer = OneHotEncoder(drop='first')

preprocess = ColumnTransformer(
    [
        ('oneHotEncoder',categorical_transformer,categorical_feature),
        ('standscler',Numeracl_transformer,numerical_feature)
    ]
)
```

```python
X =preprocess.fit_transform(X)
```

```python
pd.DataFrame(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression ,Ridge,Lasso
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
def Evel_predict(true,predict):
    r2_scor = r2_score(true,predict)
    mae= mean_absolute_error(true,predict)
    mse= mean_squared_error(true,predict)
    rmse = np.sqrt(mse)
    return r2_scor,mae,rmse
```

```python
models={
    'linearRegression':LinearRegression(),
    'SVR':SVR(),
    'KNeighborsRegressor':KNeighborsRegressor(),
    'DecisionTreeRegressor':DecisionTreeRegressor(),
    'GradientBoostingRegressor':GradientBoostingRegressor(),
    'RandomForestRegressor':RandomForestRegressor(),
    'AdaBoostRegressor':AdaBoostRegressor()
    }
for i in range(len(list(models))):
    model= list(models.values())[i]
    model.fit(X_train,y_train)
    Y_train_pred =model.predict(X_train)
    Y_test_pred = model.predict(X_test)

    
    r2_scortrain,maetrain,rmsetrain=Evel_predict(y_train,Y_train_pred)
    r2_scortest,maetest,rmsetest=Evel_predict(y_test,Y_test_pred)

    print(list(models.keys())[i])
    print('Train set prediction\n')
    print('r2_score:{:.4f}'.format(r2_scortrain))
    print('mean Absolute Error:{:.4f}'.format(maetrain))
    print('RootMeanSquaredError{:.4f}'.format(rmsetrain))

    print('-'*32)

    print('Test Set Prediction')
    print('r2_score:{:.4f}'.format(r2_scortest))
    print('mean Absolute Error:{:.4f}'.format(maetest))
    print('Root Mean Squared Error:{:.4f}'.format(rmsetest))

    print('='*32)
    print('\n')


    
    
```
