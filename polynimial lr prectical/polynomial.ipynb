```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline 

```

```python
x =6 * np.random.rand(100,1)-3
y = 0.5 * x**2 + 1.5*x+2% np.random.rand(100,1)
```

```python
plt.scatter(x,y,color='g')
plt.show()
```

## basic linear regression

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
Regression1= LinearRegression()
Regression1.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score
score = r2_score(y_test,Regression1.predict(x_test))
```

```python
plt.plot(x_train ,Regression1.predict(x_train),color='b')
plt.scatter(x_train,y_train)
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

# polynomial

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2,include_bias=True)
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
x_train_poly
```

```python
x_test_poly
```

```python
from sklearn.metrics import r2_score
Regression = LinearRegression()
Regression.fit(x_train_poly,y_train)
y_pred = Regression.predict(x_test_poly)
score = r2_score(y_test,y_pred)
print(score)
```

```python
print(Regression.coef_)
```

```python
print(Regression.intercept_)
```

```python
plt.scatter(x_train,Regression.predict(x_train_poly))
plt.scatter(x_train,y_train)
plt.show()
```

# With New Data

```python
x_new = np.linspace(-3,3,200).reshape(200,1)
x_new_poly = poly.transform(x_new)
```

```python
x_new_poly
```

```python
y_new = Regression.predict(x_new_poly)
plt.plot(x_new,y_new,'r-',linewidth=2,label='Predction')
plt.plot(x_train, y_train,'b.',label='New Traning Point')
plt.plot(x_test,y_test,'g.',label='Test dataset')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```
